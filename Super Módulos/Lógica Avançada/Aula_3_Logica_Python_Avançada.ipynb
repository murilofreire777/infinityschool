# Aula 3 – Condicionais e Laços de Repetição

## 1. Condicionais Avançadas
As estruturas condicionais permitem executar diferentes blocos de código com base em condições booleanas.

Python utiliza as palavras-chave `if`, `elif` e `else` para implementar essas decisões.

```
# Exemplo com if, elif e else
idade = 20

if idade < 13:
    print("Criança")
elif idade < 18:
    print("Adolescente")
else:
    print("Adulto")
```

### Condicionais aninhadas

```
nota = 8

if nota >= 7:
    if nota >= 9:
        print("Excelente!")
    else:
        print("Bom desempenho.")
else:
    print("Precisa melhorar.")
```

### Condicional em uma linha (ternário)

```
media = 6.5
status = "Aprovado" if media >= 6 else "Reprovado"
print("Resultado:", status)
```

## 2. Validação Robusta de Entradas
Entradas do usuário devem ser verificadas para evitar erros em tempo de execução. Podemos usar `try/except` ou verificações com string.

```
entrada = input("Digite um número inteiro: ")

if entrada.isdigit():
    numero = int(entrada)
    print("Número válido:", numero)
else:
    print("Entrada inválida. Tente novamente.")
```

```
# Validação com try/except
try:
    valor = float(input("Digite um número decimal: "))
    print("Valor lido:", valor)
except ValueError:
    print("Erro: entrada inválida!")
```

## 3. Controle de Fluxo: `break`, `continue`, `pass`
Essas instruções são usadas para controlar a execução dos laços de forma mais precisa.

- `break` → Interrompe o laço imediatamente
- `continue` → Pula o restante da iteração atual e vai para a próxima
- `pass` → Não faz nada; usado como placeholder

```
for i in range(5):
    if i == 2:
        continue  # pula o 2
    if i == 4:
        break     # interrompe antes de imprimir o 4
    print("Valor:", i)
```

```
# Uso de pass
for letra in "Python":
    if letra == 'h':
        pass  # ponto de inserção futura
    print("Letra:", letra)
```

## 4. Laços de Repetição Avançados
Os laços `for` e `while` permitem repetir blocos de código. O `for` é ideal para estruturas iteráveis, enquanto `while` é usado quando não sabemos o número de repetições de antemão.

```
# Exemplo com while
contador = 0
while contador < 5:
    print("Contador:", contador)
    contador += 1
```

```
# for com range decrescente
for i in range(10, 0, -2):
    print("Valor decrescente:", i)
```

### Funções úteis com laços:
- `enumerate()` → retorna pares (índice, valor)
- `zip()` → itera sobre múltiplas listas simultaneamente
- `sorted()` → retorna nova lista ordenada
- `reversed()` → inverte a ordem da sequência

```
# enumerate
nomes = ["Ana", "Bruno", "Carla"]
for i, nome in enumerate(nomes):
    print(f"{i}: {nome}")
```

```
# zip
nomes = ["Ana", "Bruno", "Carla"]
idades = [22, 35, 28]
for nome, idade in zip(nomes, idades):
    print(f"{nome} tem {idade} anos")
```

```
# sorted e reversed
valores = [5, 3, 8, 1]
print("Ordenado:", sorted(valores))
print("Reverso:", list(reversed(valores)))
```

## 5. Iteração sobre estruturas aninhadas
É comum ter que iterar sobre listas dentro de listas (ou outras estruturas compostas).

```
matriz = [[1, 2], [3, 4], [5, 6]]
for linha in matriz:
    for valor in linha:
        print(valor, end=' ')
    print()
```

```
# Dicionário com listas
alunos = {
    'Ana': [8, 9],
    'Bruno': [6, 7],
    'Carla': [9, 10]
}

for nome, notas in alunos.items():
    print(f"{nome}:")
    for nota in notas:
        print(" -", nota)
```

## 6. Comprehensions
As compreensões são formas concisas de gerar listas, conjuntos e dicionários a partir de iteráveis.

São mais eficientes e mais legíveis quando usadas corretamente.

```
# List comprehension com filtro
pares = [x for x in range(20) if x % 2 == 0]
print("Pares:", pares)
```

```
# Set comprehension para eliminar duplicatas
letras = "banana"
conjunto = {letra for letra in letras}
print("Letras únicas:", conjunto)
```

```
# Dict comprehension com transformação
quadrados = {x: x**2 for x in range(5)}
print("Dicionário de quadrados:", quadrados)
```

```
# Compreensão aninhada
matriz = [[1, 2], [3, 4]]
flatten = [num for linha in matriz for num in linha]
print("Lista achatada:", flatten)
```
