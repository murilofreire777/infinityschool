# Aula 2 – Variáveis, Operadores e Expressões

## 1. Variáveis e Escopo
Em Python, variáveis são criadas dinamicamente e não possuem um tipo fixo. O escopo define onde a variável pode ser acessada.

### Tipos de escopo:
- **Local**: dentro de uma função
- **Global**: fora de todas as funções
- **nonlocal**: em funções aninhadas, referenciando variáveis do escopo imediatamente superior

```python
x = 10  # variável global

def func():
    x = 5  # escopo local
    print("Local:", x)

func()
print("Global:", x)
```

```python
def externa():
    x = "externa"
    def interna():
        nonlocal x
        x = "interna"
    interna()
    print("Variável após chamada interna:", x)

externa()
```

## 2. Operadores
Python possui diversos operadores úteis, que podem ser agrupados em:
- **Aritméticos**: `+`, `-`, `*`, `/`, `//`, `%`, `**`
- **Relacionais**: `==`, `!=`, `<`, `>`, `<=`, `>=`
- **Lógicos**: `and`, `or`, `not`
- **Bit a bit**: `&`, `|`, `^`, `~`, `<<`, `>>`
- **Atribuição combinada**: `+=`, `-=`, `*=`, etc.

```python
a, b = 10, 3
print("Aritméticos:", a + b, a / b, a // b, a % b)
print("Relacionais:", a > b, a == b)
print("Lógicos:", a > 5 and b < 5)

# Bit a bit
print("AND:", a & b)
print("OR:", a | b)
print("XOR:", a ^ b)
print("Shift esquerdo:", a << 1)
print("Shift direito:", a >> 1)

# Atribuição combinada
a += 2
print("A após += 2:", a)
```

```python
print(10 ^ 5) # 15
print(bool('1010') ^ bool('0101')) # False
print(10 and 5) # 5
```

## 3. Avaliação Preguiçosa e Curto-Circuito
Avaliação preguiçosa ocorre quando uma expressão lógica é interrompida assim que o resultado é determinado.
### `and` para ao encontrar `False`, `or` para ao encontrar `True`.

```python
def foo():
    print("Executando foo()")
    return True

print("AND curto-circuito:")
print(False and foo())  # foo() não será executada

print("OR curto-circuito:")
print(True or foo())  # foo() não será executada
```

## 4. Expressões Complexas
Quando múltiplos operadores são usados, a **ordem de precedência** determina a avaliação da expressão. É sempre recomendável usar parênteses para tornar a leitura mais clara.

```python
resultado = (3 + 4) * 2 > 10 or (5 < 3)
print("Resultado:", resultado)
```

## 5. Expressões Lambda e Funções Anônimas
Expressões `lambda` permitem criar funções curtas e anônimas.
São frequentemente usadas com funções como `sorted`, `map`, `filter`, etc.

```python
# Função tradicional
def quadrado(x):
    return x * x

# Lambda
quad = lambda x: x * x
print("Lambda quadrado de 5:", quad(5))
```

```python
# Ordenando lista ignorando maiúsculas
nomes = ['ana', 'Carlos', 'bruno']
print("Ordenado:", sorted(nomes, key=lambda nome: nome.lower()))
```

```python
# Usando lambda com map
numeros = [1, 2, 3, 4]
dobrados = list(map(lambda x: x * 2, numeros))
print("Dobrados:", dobrados)
```
