# Aula 4 – Algoritmos de Ordenação e Busca

## 1. Busca Linear
A busca linear (ou sequencial) percorre todos os elementos de uma lista até encontrar o alvo. É simples, mas ineficiente para listas grandes.

### Complexidade: O(n)

```
def busca_linear(lista, alvo):
    for i, valor in enumerate(lista):
        if valor == alvo:
            return i
    return -1

dados = [4, 7, 1, 9, 2]
print("Índice do 9:", busca_linear(dados, 9))
print("Índice do 5 (não existe):", busca_linear(dados, 5))
```

## 2. Busca Binária
A busca binária só funciona com **listas ordenadas**. Ela divide a lista ao meio repetidamente.

### Complexidade: O(log n)

```
def busca_binaria(lista, alvo):
    inicio, fim = 0, len(lista) - 1
    while inicio <= fim:
        meio = (inicio + fim) // 2
        if lista[meio] == alvo:
            return meio
        elif lista[meio] < alvo:
            inicio = meio + 1
        else:
            fim = meio - 1
    return -1

ordenada = [1, 3, 5, 7, 9]
print("Índice do 7:", busca_binaria(ordenada, 7))
print("Índice do 2 (não existe):", busca_binaria(ordenada, 2))
```

## 3. Bubble Sort
Algoritmo simples que compara pares de elementos adjacentes e os troca se estiverem fora de ordem.

### Complexidade: O(n²)

```
def bubble_sort(lista):
    n = len(lista)
    for i in range(n):
        for j in range(n - 1 - i):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

valores = [5, 3, 8, 1]
bubble_sort(valores)
print("Ordenado (bubble):", valores)
```

## 4. Insertion Sort
Constrói a lista ordenada um elemento por vez, inserindo cada novo valor na posição correta.

### Complexidade: O(n²)

```
def insertion_sort(lista):
    for i in range(1, len(lista)):
        chave = lista[i]
        j = i - 1
        while j >= 0 and lista[j] > chave:
            lista[j + 1] = lista[j]
            j -= 1
        lista[j + 1] = chave

valores = [5, 2, 4, 6]
insertion_sort(valores)
print("Ordenado (insertion):", valores)
```

## 5. Selection Sort
Seleciona o menor elemento da lista e o coloca na posição correta repetidamente.

### Complexidade: O(n²)

```
def selection_sort(lista):
    n = len(lista)
    for i in range(n):
        menor = i
        for j in range(i + 1, n):
            if lista[j] < lista[menor]:
                menor = j
        lista[i], lista[menor] = lista[menor], lista[i]

valores = [29, 10, 14, 37]
selection_sort(valores)
print("Ordenado (selection):", valores)
```

## 6. Ordenação em Python
Python fornece funções internas eficientes para ordenação:
- `sorted(lista)` → retorna nova lista ordenada
- `lista.sort()` → ordena a própria lista
- Argumento `key` permite ordenações personalizadas

```
nomes = ["Carlos", "ana", "Bruno"]
print("Ordenado (case sensitive):", sorted(nomes))
print("Ordenado (case insensitive):", sorted(nomes, key=str.lower))
```

```
# Ordenando tuplas por segundo elemento
dados = [(1, 'b'), (3, 'a'), (2, 'c')]
ordenado = sorted(dados, key=lambda x: x[1])
print("Ordenado por letra:", ordenado)
```

## 7. Comparação de Algoritmos
| Algoritmo       | Complexidade Média | Estável | Simples? |
|----------------|--------------------|---------|----------|
| Bubble Sort    | O(n²)              | Sim     | Sim      |
| Insertion Sort | O(n²)              | Sim     | Sim      |
| Selection Sort | O(n²)              | Não     | Sim      |
| Busca Linear   | O(n)               | –       | Sim      |
| Busca Binária  | O(log n)           | –       | Sim (com ordenação)

**Observação**: Para grandes volumes de dados, o ideal é utilizar os métodos internos do Python, que são baseados no algoritmo Timsort (eficiente e estável).
