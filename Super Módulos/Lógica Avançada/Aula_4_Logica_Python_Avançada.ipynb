## 1. Funções: Definição e Parâmetros
Funções são blocos de código reutilizáveis que recebem entradas (parâmetros), processam e retornam uma saída. Vamos rever os diferentes tipos de parâmetros em funções Python.

```python
def saudacao(nome, mensagem="Olá"):
    print(f"{mensagem}, {nome}!")

saudacao("Maria")
saudacao("Carlos", "Bem-vindo")

#Função com documentação
def buildConnectionString(params):
    """Build a connection string from a dictionary of parameters.

    Returns string."""
```

### Argumentos variáveis com *args e **kwargs
Permitem passar quantidades indefinidas de argumentos posicionais (`*args`) ou nomeados (`**kwargs`).

```python
def somar(*numeros):
    return sum(numeros)

print("Soma:", somar(1, 2, 3, 4))
```

```python
def mostrar_dados(**info):
    for chave, valor in info.items():
        print(f"{chave}: {valor}")

mostrar_dados(nome="João", idade=30, cidade="Salvador")
```

## 2. Retorno de Múltiplos Valores
Funções podem retornar múltiplos valores em forma de tupla.

```python
def calcular(valores):
    soma = sum(valores)
    media = soma / len(valores)
    return soma, media

s, m = calcular([10, 20, 30])
print("Soma:", s)
print("Média:", m)
```

## 3. Funções Aninhadas e Closures
Funções podem ser definidas dentro de outras. Closures ocorrem quando a função interna mantém acesso às variáveis da função externa.

```python
def saudacao_criador(mensagem):
    def saudar(nome):
        print(f"{mensagem}, {nome}!")
    return saudar

boa_noite = saudacao_criador("Boa noite")
boa_noite("Maria")
```

## 4. Funções como Argumentos e Retorno
Funções são objetos de primeira classe e podem ser passadas como argumentos ou retornadas por outras funções.

```python
def aplicar(funcao, lista):
    return [funcao(x) for x in lista]

def quadrado(x):
    return x * x

print(aplicar(quadrado, [1, 2, 3, 4]))
```

```python
def multiplicador(fator):
    def multiplicar(x):
        return x * fator
    return multiplicar

dobro = multiplicador(2)
print(dobro(10))
```

## 5. Recursão
Recursão ocorre quando uma função chama a si mesma. Muito usada em algoritmos como fatorial, fibonacci, árvores.

```python
def fatorial(n):
    if n <= 1:
        return 1
    return n * fatorial(n - 1)

print("Fatorial de 5:", fatorial(5))
```

## 6. Funções Lambda

Funções Lambda são funções de 1 única linha. Nunca são exigidas. Somente permite comandos de 1 linha.

```python
def f(x):
    return x*2
print(f(3))

g = lambda x: x*2
print(g(3))

print( (lambda x: x*2)(3) ) #Função sem atribuição
```

## 6. Introdução a Decoradores
Decoradores são funções que modificam o comportamento de outras funções.

```python
def log(func):
    def wrapper(*args, **kwargs):
        print(f"Chamando função {func.__name__} com args={args}")
        return func(*args, **kwargs)
    return wrapper

@log
def somar(a, b):
    return a + b

print("Resultado:", somar(2, 3))
```

## 7. Dunders Úteis: Criando uma Estrutura com `__iter__`
Dunder methods (Double Underscore Methods) são usados para implementar comportamentos especiais de objetos.

```python
class No:
    def __init__(self, valor):
        self.valor = valor
        self.proximo = None

class ListaEncadeada:
    def __init__(self):
        self.inicio = None

    def adicionar(self, valor):
        novo = No(valor)
        if not self.inicio:
            self.inicio = novo
        else:
            atual = self.inicio
            while atual.proximo:
                atual = atual.proximo
            atual.proximo = novo

    def __iter__(self):
        atual = self.inicio
        while atual:
            yield atual.valor
            atual = atual.proximo

lista = ListaEncadeada()
lista.adicionar(1)
lista.adicionar(2)
lista.adicionar(3)

for item in lista:
    print(item)
```

## 8. Outros Dunders Úteis
Exemplos de `__str__`, `__len__`, `__eq__`, etc.

```python
class Pessoa:
    def __init__(self, nome):
        self.nome = nome

    def __str__(self):
        return f"Pessoa: {self.nome}"

    def __len__(self):
        return len(self.nome)

    def __eq__(self, outro):
        return self.nome == outro.nome

p1 = Pessoa("Joana")
p2 = Pessoa("Joana")
print(p1)
print("Tamanho do nome:", len(p1))
print("p1 == p2?", p1 == p2)
```
