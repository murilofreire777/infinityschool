## 1. Sistemas Numéricos


![Conversa entre o ET e o Astronauta](piada_et.jpg)

Python permite trabalhar com diferentes bases numéricas: binária, octal, decimal e hexadecimal. 
Isso é útil para aplicações como codificação de dados, endereçamento de memória, redes, e lógica digital.

### Conversões entre sistemas:
- `bin(x)` → retorna representação binária
- `oct(x)` → retorna representação octal
- `hex(x)` → retorna representação hexadecimal
- `int(string, base)` → converte de qualquer base para decimal


```python
x = 42
print("Decimal:", x)
print("Binário:", bin(x))
print("Octal:", oct(x))
print("Hexadecimal:", hex(x))

# Convertendo de string binária e hexadecimal para inteiro
print("Decimal de '101010' (binário):", int('101010', 2))
print("Decimal de '2A' (hexadecimal):", int('2A', 16))
```

## 2. Representação em Memória das Variáveis

Em Python, variáveis são **nomes que referenciam objetos** na memória. 
A função `id()` retorna o identificador único do objeto (geralmente ligado ao endereço de memória).

Objetos **imutáveis** (como `int`, `float`, `str`, `tuple`) não podem ser modificados após a criação.
Objetos **mutáveis** (como `list`, `dict`, `set`) podem ser alterados diretamente.


```python
a = 10
a = 5
# O valor 10 deixa de ser referenciado, um novo objeto é criado.
# Se não houver referências para este valor, o coletor de lixo eliminará o objeto da memória.
```

```python
a = 10
b = 10
print("a is b:", a is b)  # Aponta para o mesmo objeto
print("id(a):", id(a), "| id(b):", id(b))

lista1 = [1, 2, 3]
lista2 = lista1
lista2.append(4)
print("lista1:", lista1)
print("lista2:", lista2)
```

Diferenças de representação de um número

```python
# Uma coisa é a representação de um número (em binário ou decimal);
# outra coisa é a maneira como o Python mostra esse valor
a = 0b101
b = 0b111
c = a+b
print(c) 

a = 0b101
b = 0b111
c = bin(a+b)
print(c)
```

### Cópias rasas vs profundas

```python
import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)
deep = copy.deepcopy(original)

original[0][0] = 999
print("Original:", original)
print("Shallow:", shallow)
print("Deep:", deep)
```

## ⚠️ Problemas com Ponto Flutuante (float)

Em Python, números do tipo `float` são representados internamente usando a padronização IEEE 754, que usa binário para representar números decimais. 
Por causa disso, alguns valores aparentemente simples como `0.1` e `0.2` não possuem representação binária exata.

Esse problema pode causar resultados inesperados em somas e comparações.

### Exemplo:

```python
#Imprecisão 
print(0.1 + 0.2)         # Esperado: 0.3
print((0.1 + 0.2) == 0.3)  # Esperado: True
```

```python
#Efeitos em loops
total = 0.0
for _ in range(10_000):
    total += 0.0001
print(total)  # pode não dar exatamente 1.0
```

Transformar 0.2 em binário
1. Colocar 0.____ <br><br>
2. Multiplicar por 2. A parte após a virgula, integra o resultado<br>
0.2 x 2 == 0.4<br>
No resultado fica 0.0___ e sobra 0.4<br><br>
3.Repete para o novo resultado<br>
0.4 x 2 = 0.8<br>
No resultado fica 0.00__ e sobra 0.8<br>
0.8 x 2 = 1.6<br>
No resultado fica 0.001__ e sobra 0.6<br>
0.6 x 2 = 1.2<br>
No resultado fica 0.0011___ e sobra 0.2<br>
0.2 x 2 = 0.4...<br><br>
Isso gerará uma dízima periódica


### ✅ Solução: usar o módulo `decimal`
O módulo `decimal` permite trabalhar com precisão decimal exata, útil especialmente em aplicações financeiras.

```python
from decimal import Decimal

a = Decimal('0.1')
b = Decimal('0.2')
c = a + b

print("Com Decimal:", c)
print("Comparação exata:", c == Decimal('0.3'))
```

## 3. Tipos Numéricos Especiais

Python oferece tipos adicionais além de `int` e `float`:

- `Decimal` → usado quando se deseja **alta precisão**, como em cálculos financeiros
- `Fraction` → representação exata de frações
- `complex` → números complexos (parte real e imaginária)


```python
from decimal import Decimal
from fractions import Fraction

# Decimal
print("Decimal:", Decimal('0.1') + Decimal('0.2'))

# Fraction
print("Fraction:", Fraction(3, 4) + Fraction(1, 4))

# Complex
z = 2 + 3j
print("Complexo:", z)
print("Parte real:", z.real)
print("Parte imaginária:", z.imag)
```

## 4. Strings e Tuplas

**Strings** são imutáveis — operações que parecem modificar a string na verdade retornam uma nova string.
**Tuplas** são semelhantes às listas, porém também imutáveis.

```python
texto = "Python"
print(texto.upper())
print(texto[0:3])
print('-'.join(texto))
print(texto.replace('Py', 'Ja'))
```

```python
tupla = (10, 20, 30)
a, b, c = tupla
print(f"a={a}, b={b}, c={c}")
print("Primeiro elemento:", tupla[0])
```

## 5. Anotações de Tipo e Desempacotamento

```python
from typing import List

def dobrar_lista(lista: List[int]) -> List[int]:
    return [x * 2 for x in lista]

print(dobrar_lista([1, 2, 3]))
```

```python
valores = [1, 2, 3, 4, 5]
x, y, *resto = valores
print("x:", x)
print("y:", y)
print("resto:", resto)
```
